```python
import pandas as pd
import numpy as np

df = pd.read_csv("Travel details dataset.csv")

print(df.head())
```

```python
print(df.isnull().sum())
```

```python
missing_values = df.isnull().sum()

print("Missing Values:")
display(missing_values)
```

```python
print("Duplicate rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicate rows after removal:", df.duplicated().sum())

print(df.head())
```

```python
df["Start date"] = pd.to_datetime(
    df["Start date"],
    errors="coerce"
)

df["End date"] = pd.to_datetime(
    df["End date"],
    errors="coerce"
)

print(df[["Start date", "End date"]].head())
```

```python
df["Accommodation cost"] = pd.to_numeric(
    df["Accommodation cost"],
    errors="coerce"
)

df["Transportation cost"] = pd.to_numeric(
    df["Transportation cost"],
    errors="coerce"
)

print(df[
    [
        "Accommodation cost",
        "Transportation cost"
    ]
].head())
```

```python
print(df.info())
```

```python
numeric_columns = [
    "Duration (days)",
    "Traveler age",
    "Accommodation cost",
    "Transportation cost"
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = [
    "Destination",
    "Traveler gender",
    "Traveler nationality",
    "Accommodation type",
    "Transportation type"
]

for col in categorical_columns:
    df[col] = df[col].fillna("Unknown")
```

```python
df["Start date"] = df["Start date"].fillna(
    df["Start date"].mode()[0]
)

df["End date"] = df["End date"].fillna(
    df["End date"].mode()[0]
)
```

```python
df["Total Trip Cost"] = (
    df["Accommodation cost"] +
    df["Transportation cost"]
)

display(
    df[
        [
            "Trip ID",
            "Destination",
            "Accommodation cost",
            "Transportation cost",
            "Total Trip Cost"
        ]
    ].head()
)
```

```python
Q1 = df["Total Trip Cost"].quantile(0.25)

Q3 = df["Total Trip Cost"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Total Trip Cost"] < lower) |
    (df["Total Trip Cost"] > upper)
]

print("Total Trip Cost Outliers:")

display(
    outliers[
        [
            "Trip ID",
            "Destination",
            "Duration (days)",
            "Total Trip Cost"
        ]
    ]
)
```
